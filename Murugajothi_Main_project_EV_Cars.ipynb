```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)
```

```python
print("Duplicate Rows Before Cleaning:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
numeric_columns = df.select_dtypes(include="number").columns

print("Outlier Count:")

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_limit) |
                (df[col] > upper_limit)).sum()

    print(col, ":", outliers)
```

```python
numeric_columns = df.select_dtypes(include="number").columns

print("Skewness:")
print(df[numeric_columns].skew())
```

```python
import numpy as np

df["log_motor_power"] = np.log1p(df["motor_power_kW"])
df["log_avg_speed"] = np.log1p(df["avg_driving_speed_kmh"])

print("Transformed Features Created:")
print(df[["motor_power_kW", "log_motor_power",
          "avg_driving_speed_kmh", "log_avg_speed"]].head())
```

## **Analysis of EV Battery Capacity**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.histplot(df["battery_capacity_kWh"], bins=20, kde=True)

plt.title("Distribution of EV Battery Capacity")
plt.xlabel("Battery Capacity (kWh)")
plt.ylabel("Number of EVs")

plt.show()
```

### **Insights**

• Most EVs have battery capacities roughly between 40 and 80 kWh.

• The highest concentration of EVs is around 50–65 kWh.

• Very high battery capacities above 90 kWh are less common.

• The dataset contains EVs with battery capacities approximately from 15 to 110 kWh.
